There is $n$ samples and $d$ dimension (features)

Given a feature so a column $ j \in \{0, ..., d-1\} $, we want to get mean and std and do (notation: $\vec{\alpha_i} = \mathbf{x}(i,j) \ \forall j$) $$\frac{\alpha_i - \overline{\alpha_i}}{\sigma_{\alpha_i}} $$

```python
def normalize(x):
    shape = x.shape
    if len(shape) != 2:
        print("argument is not n times d matrix")
        return
    else:
        n, d = x.shape
    # create mean matrix and std matrix
    mean_x = np.mean(x, axis=0)
    mean_x = (np.ones([n, 1])) @ mean_x.reshape(1, d)
    std_x = np.std(x, axis=0)
    # print(std_x)
    std_x = (np.ones([n, 1])) @ std_x.reshape(1, d)

    norm_x = (x - mean_x) / std_x
    return norm_x
```

```python
# better version:
# it's called standardization, I should raise an error instead, I can use directly the (d,) shape
# of mean and std, and I should be carefull with std=0
def standardize(x):
    if len(x.shape) != 2:
        raise ValueError("argument is not n times d matrix")

    std = np.std(x, axis=0)
    std = np.where(std == 0, 1, std)
    return (x - np.mean(x, axis=0)) / std
```
